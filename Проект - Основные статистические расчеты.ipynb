```python
df_for_platform = df[['Platform', 'Name']].groupby('Platform', as_index=False).count().sort_values('Name', ascending = False)
df_for_platform
```

```python
znam = df_for_platform.Name.sum() 
```

```python
df_for_platform['chastota'] = round(df_for_platform.Name/znam * 100, 2)
df_for_platform_new = df_for_platform.query('chastota > 7')
df_for_platform_new
```

```python
df_for_platform_new.Platform.to_list()
```

6. Визуализируйте динамику изменения объема мировых продаж (Global_Sales) по годам для игр Nintendo следующих жанров: Fighting, Simulation, Platform, Racing, Sports.

```python
df_my = df[df['Publisher'] == 'Nintendo']
```

```python
df_my = df_my.query("Genre in ('Fighting', 'Simulation', 'Platform', 'Racing', 'Sports')").groupby(['Genre', "Year"], 
                                                                                           as_index = False).sum()
df_my
```

```python
plt.figure(figsize = (16,7))
sns.lineplot (x = 'Year', y = 'Global_Sales', hue = 'Genre', data = df_my)
```
